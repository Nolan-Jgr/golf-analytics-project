## 5. Additional Variables

Additional variables will be created from the existing data to make the dataset more useful for analysis.

Score to par

```python
df["ScoreToPar"] = df["Score"] - df['CoursePar']
```

Birdie Percentage

```python
df["BirdiePct"] = df["Birdies"] / 18
```

Bogey Percentage

```python
df["BogeyPct"] = df["Bogeys"] / 18
```

Earnings per Stroke

```python
df["EarningsPerStroke"] = df["Earnings"] / df["Score"]
```
