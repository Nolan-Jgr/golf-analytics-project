### Feature Engineering

#### Score to Par

```python
df["ScoreToPar"] = df["Score"] - df["CoursePar"]
```

#### Birdie Percentage

```python
df["BirdiePct"] = df["Birdies"] / 18
```

#### Bogey Percentage

```python
df["BogeyPct"] = df["Bogeys"] / 18
```

#### Earnings per Stroke

```python
df["EarningsPerStroke"] = df["Earnings"] / df["Score"]
```

Several derived variables were created to make the dataset more useful for
later analysis.

- **ScoreToPar** measures performance relative to the course par.
- **BirdiePct** represents birdies per hole.
- **BogeyPct** represents bogeys per hole.
- **EarningsPerStroke** provides a normalized measure of tournament earnings
  relative to the number of strokes recorded.

### Optional Attribute(s)

```python
fairway_pct = (df["FairwaysHit"].mean() / 14) * 100
print("Average Fairway Percentage:", round(fairway_pct, 2), "%")
```

```python
gir_pct = (df["GIR"].mean() / 18) * 100
print("Average Greens in Regulation Percentage:", round(gir_pct, 2), "%")
```
